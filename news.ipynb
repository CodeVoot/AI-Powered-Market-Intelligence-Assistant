<h3>Test Stock Data </h3>
<p> WE are going to find the data based on their ticker symbol like COLCAP AAPL </p>

```python
def get_stock_data(symbol):
    stock = yf.Ticker(symbol) 
    data = stock.history(period="5d") #last 5days data
    latest = data["Close"].iloc[-1] # most recent closing price 
    prev = data["Close"].iloc[-2] # the previous day's closing price

    return{
        "price": str(round(float(latest),2)) + "$",

        "change": round(float(latest-prev),2),

        "percent_change": round(float((latest-prev)/prev*100),2)
    }
    
```

```python
get_stock_data("BBNX")
```

<h3>Get Company News </h3>

```python
def get_company_news(company):
    url = f"https://newsapi.org/v2/everything?q={company}&pageSize=5&apiKey={NEWS_API_KEY}"
    response = requests.get(url)
    data = response.json()

    news=[]

    for article in data["articles"]:
        news.append({

            "title":article["title"],

            "description":article["description"]

        })

    return news
```

```python
news = get_company_news("Apple")

news
```
